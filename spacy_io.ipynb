```python

import spacy
nlp = spacy.load("en_core_web_sm")
texto = "Eu sou o framework spaCy."
doc = nlp(texto)
for token in doc:
  print(token.texto)

for token in doc:
  print(token.text, token.pos_)

for ent in doc.ents:
  print(ent.text, ent.label_)

```
